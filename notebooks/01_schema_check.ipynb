```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/marketing_AB.csv")

print("SHAPE:", df.shape)
print("\nCOLUMNS:", list(df.columns))
print("\nDTYPES:\n", df.dtypes)
print("\nHEAD:\n", df.head().to_string())
print("\nNULLS:\n", df.isna().sum())
print("\nDUPLICATE user id:", df["user id"].duplicated().sum())

g = df.groupby("test group")["converted"].agg(["count", "sum", "mean"])
g.columns = ["n_users", "n_converted", "conversion_rate"]
print("\nGROUPS:\n", g.to_string())

# PSA stability — dihitung, bukan diasumsikan
for grp in g.index:
    n, p = g.loc[grp, "n_users"], g.loc[grp, "conversion_rate"]
    se = np.sqrt(p * (1 - p) / n)
    print(f"\n{grp}: n={n}, p={p:.5f}, SE={se:.5f}, "
          f"95% CI=[{p - 1.96*se:.5f}, {p + 1.96*se:.5f}], width={2*1.96*se:.5f}")

# balance check
for col in ["most ads day", "most ads hour"]:
    print(f"\nBALANCE — {col}:\n",
          pd.crosstab(df[col], df["test group"], normalize="columns").round(4).to_string())

print("\nEXPOSURE — total ads by group:\n",
      df.groupby("test group")["total ads"].describe().to_string())
```
