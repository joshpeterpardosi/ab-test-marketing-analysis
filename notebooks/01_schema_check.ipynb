# Phase 1 — Schema Check

Dataset: `marketing_AB.csv` (Kaggle, faviovaz), ad vs PSA.

Purpose: establish what the data actually contains before any test is run —
shape, types, missingness, key uniqueness, group sizes, and whether the PSA arm
is large enough to compare against. Nothing here is a hypothesis test.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/marketing_AB.csv")

print("SHAPE:", df.shape)
print("\nCOLUMNS:", list(df.columns))
print("\nDTYPES:\n", df.dtypes)
print("\nHEAD:\n", df.head().to_string())
```

## Integrity — missingness and key uniqueness

One row per user is assumed throughout the analysis. A duplicate `user id` would
break that assumption, so it is checked rather than trusted.

```python
print("NULLS:\n", df.isna().sum())
print("\nDUPLICATE user id:", df["user id"].duplicated().sum())
```

## Group sizes and PSA stability

The design is heavily imbalanced toward the ad arm, so the question is whether
the PSA estimate is precise enough to compare against at all. The interval below
is computed rather than assumed: if its width were comparable to the gap being
tested, the comparison would not be usable.

```python
g = df.groupby("test group")["converted"].agg(["count", "sum", "mean"])
g.columns = ["n_users", "n_converted", "conversion_rate"]
print("GROUPS:\n", g.to_string())

for grp in g.index:
    n, p = g.loc[grp, "n_users"], g.loc[grp, "conversion_rate"]
    se = np.sqrt(p * (1 - p) / n)
    print(f"\n{grp}: n={n}, p={p:.5f}, SE={se:.5f}, "
          f"95% CI=[{p - 1.96*se:.5f}, {p + 1.96*se:.5f}], width={2*1.96*se:.5f}")
```

## Balance and exposure

`most ads day`, `most ads hour` and `total ads` are post-treatment — they
describe exposure behaviour after assignment, not covariates measured before it.
The comparison below is therefore a diagnostic to size the imbalance, not a
randomization check. Phase 2 tests it formally.

```python
for col in ["most ads day", "most ads hour"]:
    print(f"\nBALANCE — {col}:\n",
          pd.crosstab(df[col], df["test group"], normalize="columns").round(4).to_string())

print("\nEXPOSURE — total ads by group:\n",
      df.groupby("test group")["total ads"].describe().to_string())
```
